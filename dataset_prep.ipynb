```python
import pandas as pd
from sbd import sentencePipeline
```

```python
df = pd.read_csv("IndianFinancialNews.csv")
```

```python
df.Description
```

```python
all_sentences = []

for text in df['Description'].dropna():
    sentences = sentencePipeline(str(text))  # ensure string
    all_sentences.extend(sentences)

with open("output_sentences.txt", "w", encoding="utf-8") as f:
    for sentence in all_sentences:
        f.write(sentence.strip() + "\n")
```
